```python
from langchain_ollama import ChatOllama
```

```python
llm = ChatOllama(
    model='qwen2.5:3b',
    temperature=0
)
```

```python
prompts = [
    {
        'role':'user',
        'content':'my name is krati beohar'
    },
    {
       'role':'ai','content':"Hello Krati Beohar! It's nice to meet you. How can I assist you today? Do you have any specific questions or topics that you'd like to discuss?"
    },
    {
        'role':'user','content':'what is my name?'
    }

]

res = llm.invoke(prompts)
res.content
```

#### manage history


```python
history = []
while True:
    query = input("user: ")
    if query.lower() in ['exit', 'bye', 'quit']:

        
        print('good bye🦋')
    history.append({'role':'user','content':query})
    print("user: ", query)

    res = llm.invoke(history)
    history.append({'role':'ai','content':res.content})

    print("AI: ",res.content,'\n')
```
