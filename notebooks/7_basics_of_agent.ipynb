```python
from langchain_ollama import ChatOllama
```

```python
from langchain_core.tools import tool
from langchain.agents import create_agent
```

```python
@tool
def add_number(a:int,b:int):
    """
      this function return addition of two number
      Args:
         a : number1
         b : number2
    """
    return a+b
@tool
def sub_number(a:int,b:int):
    """
      this function return subtraction of two number
      Args:
         a : number1
         b : number2
    """
    return a-b
```

```python
add_number.invoke({'a':12, 'b':13})
```

```python
sub_number.invoke({'a':19, 'b':13})
```

```python
llm = ChatOllama(
    model = 'qwen2.5:3b'
)
```

```python
agent  = create_agent(
    model = llm,
    tools = [add_number, sub_number],
    system_prompt="you are a math teacher which use tool for calculation like adding or subtracting, be helpful"
)
```

```python
response = agent.invoke({"messages":[{"role":"user",'content':'could you add 3+5'}]})
response['messages'][-1].content
```
