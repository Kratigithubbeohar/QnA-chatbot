####wikipedia loader

```python
from langchain_community.document_loaders import WikipediaLoader
import wikipedia

wikipedia.set_user_agent(
    "MyGenAIProject/1.0 (contact: your-email@example.com)"
)

loader = WikipediaLoader(
    query="Generative AI",
    load_max_docs=2
)

docs = loader.load()

print(len(docs))

for doc in docs:
    print(doc.metadata)
    print(doc.page_content[:500])
```
