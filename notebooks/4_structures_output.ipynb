```python
from langchain_ollama import ChatOllama
```

```python
llm = ChatOllama(
    model = 'qwen2.5:3b',
    temperature=0
)
```

```python
text = "hellow, I'm Krati Beohar, email is kratibeohar@gmail.com and my age is 21"
```

```python
res = llm.invoke(f"please give me only name email of user {text}")
```

```python
res.content
```

```python
res.name
```

```python
res.email
```

####structured output

```python
from pydantic import BaseModel, Field
```

```python
class response(BaseModel):
    name:str = Field(description="name of the user")
    email:str = Field(description= "email of the user")
    age:int = Field(description="age of user")
```

```python
structured_llm = llm.with_structured_output(response)
```

```python
res = structured_llm.invoke(f"please give me only name email of user {text}")
```
